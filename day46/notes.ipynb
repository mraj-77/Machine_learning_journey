# Curse of Dimensionality
1. Dimensionality
- Dimensionality refers to the number of features (columns) in a dataset.
- 1 feature = 1 dimension.
- 10 features = 10 dimensions.

# 2. Curse of Dimensionality
- Definition: Problems that occur when the number of features (dimensions) increases in a dataset.
- As dimensions increase, data becomes sparse and finding meaningful patterns becomes difficult.
- It can reduce ML model performance.

# 3. Why Does It Happen?
1. Data Sparsity
- Data points become more spread out as dimensions increase.
- Same amount of data occupies a larger feature space.

2. Distance Concentration
- Distances between data points become increasingly similar in high-dimensional space.
- Finding nearest neighbors becomes difficult.
- Mainly affects KNN and clustering algorithms.
3. More Data Required
- Higher dimensions require significantly more training observations to represent the feature space properly.
Example:
Dimensions	Regions (10 intervals per feature)
1D	10
2D	100
3D	1,000
4D	10,000
5D	100,000


Formula: \(10^d\), where d = number of dimensions.

# 4. Effects on ML Algorithms
- KNN: Distance becomes less meaningful.
- K-Means: Finding meaningful clusters becomes difficult.
- Linear Regression: Can suffer from overfitting and multicollinearity.
- SVM: Performance may decrease in high-dimensional data.
- Decision Tree: Can overfit with unnecessary features.

# 5. Problems Caused
1. Overfitting.
2. Increased computational cost.
3. Data sparsity.
4. Distance becomes less meaningful.
5. More training data required.

# 6. How to Handle Curse of Dimensionality?
1. Feature Selection
- Select important features.
- Remove irrelevant or unnecessary columns.
2. PCA (Principal Component Analysis)
- Converts original features into fewer new features called principal components.
- Preserves as much variance as possible.
3. Regularization
- Reduces overfitting by penalizing large model coefficients.
- Types: L1 (Lasso) and L2 (Ridge).
- L1 can also perform feature selection.
4. Collect More Data
- Increase training observations to better represent the feature space.
5. Domain Knowledge
- Select features based on real-world understanding.

# 7. Feature Selection vs PCA
Feature Selection	PCA
Selects existing features.	Creates new features.
Removes unnecessary columns.	Combines original feature information.
Original features remain.	New components may be difficult to interpret.
Example: SelectKBest	Example: PCA


# 8. Important Points
- More features do not always mean better accuracy.
- High dimensions can increase overfitting.
- KNN is sensitive to the curse of dimensionality.
- PCA is a popular dimensionality reduction technique.
- Feature selection removes irrelevant features.

# 9. Interview Definition
Curse of Dimensionality: It refers to problems such as data sparsity, distance concentration, and increased computational complexity that arise as the number of features increases.
