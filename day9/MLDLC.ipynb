# 1. Frame the Problem

Sabse pehle samjho ki problem kya hai aur ML se kya solve karna hai.

Kya predict karna hai? (price, spam, fraud)
Success kaise napenge?
Kya ML sach mein chahiye?

Example: “Customer app chhodne wala hai ya nahi, ye predict karna.”

# 2. Gathering Data

Problem ke liye data ikattha karo. Sources: API, web scraping, database, surveys, files (CSV).

# 3. Data Preprocessing

Raw data ganda hota hai, use saaf karna padta hai.

Missing values bharna
Duplicates hatana
Galat values theek karna
Data ko sahi format mein badalna (text → numbers)

# 4. Exploratory Data Analysis (EDA)

Data ko graphs aur statistics se samjho.

Kaun se columns important hain?
Koi pattern ya trend dikh raha hai?
Outliers (ajeeb values) hain kya?

# 5. Feature Engineering and Selection

Features = data ke columns jinse model seekhta hai.

Engineering: Naye useful columns banana. Jaise “date of birth” se “age” nikalna.
Selection: Faltu columns hata dena (jaise “malik ka favourite color”).

# 6. Model Training, Evaluation and Selection
Training: Alag-alag algorithms ko data par seekhne dena
Evaluation: Accuracy check karna (test data par)
Selection: Jo model sabse accha chale, use chunna

# 7. Model Deployment

Chune hue model ko asli duniya mein laga dena, jaise website, app ya server par, taaki log use kar sakein.

# 8. Testing

Deploy hone ke baad real users ke saath check karo ki sab sahi chal raha hai ya nahi. Koi bug ya galat predictions to nahi?

# 9. Optimize

Model ko lagataar behtar banate raho.

Naya data add karo
Model dobara train karo
Speed aur accuracy improve karo

Ye isliye zaroori hai kyunki duniya badalti rehti hai (Day 7 ke “Offline Learning” challenge se yaad karo).
