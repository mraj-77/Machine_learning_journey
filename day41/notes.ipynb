# 1. What are Outliers?
Definition: An Outlier is a data point that is significantly different from the majority of observations in a dataset.

In simple words, values that are unusually high or unusually low compared to other values are called Outliers.

Example:

Suppose we have students' marks:
marks = [45, 50, 48, 52, 47, 49, 51, 98]


Here, 98 is unusually high compared to other marks.

So, 98 may be an outlier.

Real-life Examples:
- Student marks: Most students score 40–60, but one scores 100.
- House prices: Most houses cost ₹20–50 lakh, but one costs ₹10 crore.
- Salary: Most employees earn ₹20,000–50,000, but one earns ₹10 lakh.
- Age: Most customers are 20–50 years old, but one record shows 200 years.
Remember: Every unusual value is not necessarily an error.

# 2. When Are Outliers Dangerous?
Outliers become dangerous when they negatively affect statistical analysis or Machine Learning model performance.

Example:

Suppose house prices are:
[10, 12, 15, 18, 20, 1000]


Here, 1000 is an extreme value.

It can significantly increase the average house price.
When can outliers be dangerous?
1. When they are caused by data entry errors.
2. When they significantly affect the mean.
3. When they distort relationships between features.
4. When they negatively affect model predictions.
5. When they affect distance-based algorithms.
Important: Before removing an outlier, understand whether it is a genuine observation or an error.

# 4. How to Treat Outliers?
There are several techniques to handle outliers.
### 1. Trimming
- Remove outlier observations from the dataset.
- Useful when outliers are errors or irrelevant observations.
Example:
df = df[df['Age'] < 100]


### 2. Capping (Winsorization)
- Replace extreme values with a specified upper or lower limit.
- Instead of deleting data, we limit extreme values.
Example:
df['Age'] = df['Age'].clip(upper=80)


Here, all values above 80 become 80.
### 3. Transformation
- Apply mathematical transformations to reduce the effect of extreme values.
- Common transformations include:
  - Log Transformation
  - Square Root Transformation
  - Reciprocal Transformation
Example:
import numpy as npdf['Salary_log'] = np.log1p(df['Salary'])


### 4. Imputation
- Replace outliers with suitable statistical values.
- For example, median or percentile values.
Example:
median = df['Salary'].median()df.loc[df['Salary'] > 1000000, 'Salary'] = median


Note: Do not automatically replace every extreme value with the median. First investigate whether it is valid.

# 6. Statistical Techniques for Outlier Detection
A. Z-Score Method
Definition: Z-score measures how many standard deviations a data point is away from the mean.

Formula:
Z = \frac{X-\mu}{\sigma}

Where:
- X = Individual data point
- μ = Mean
- σ = Standard deviation
Rule:
- Z-score > +3 → Potential upper outlier.
- Z-score < -3 → Potential lower outlier.
- Z-score between -3 and +3 → Generally considered normal.
Python Code:
from scipy.stats import zscoredf['Z_score'] = zscore(df['Salary'])outliers = df[abs(df['Z_score']) > 3]


Important: Z-score works best when data is approximately normally distributed and can be influenced by extreme values.
B. IQR Method (Interquartile Range)
Definition: IQR measures the spread of the middle 50% of the data.

Formula:
IQR = Q3-Q1

Where:
- Q1 = 25th percentile.
- Q3 = 75th percentile.
Outlier Boundaries:
Lower = Q1-1.5 \times IQR

Upper = Q3+1.5 \times IQR

Rule:
- Values below Lower Bound → Potential outliers.
- Values above Upper Bound → Potential outliers.
Example:

Suppose:
Q1 = 20
Q3 = 40

Then:
IQR = 40 - 20 = 20

Lower Bound = 20 - (1.5 × 20)
            = -10

Upper Bound = 40 + (1.5 × 20)
            = 70

Any value below -10 or above 70 is considered a potential outlier.

Python Code:
Q1 = df['Salary'].quantile(0.25)Q3 = df['Salary'].quantile(0.75)IQR = Q3 - Q1lower = Q1 - 1.5 * IQRupper = Q3 + 1.5 * IQRoutliers = df[    (df['Salary'] < lower) |    (df['Salary'] > upper)]


Important: IQR is generally more robust than Z-score when the data is skewed.

# 7. Advanced Techniques for Outlier Detection
Technique	Working
Isolation Forest	Isolates unusual observations using random trees.
Local Outlier Factor (LOF)	Identifies observations with unusually low local density.
DBSCAN	Identifies points that do not belong to dense clusters as noise.
One-Class SVM	Learns the boundary of normal observations and identifies unusual points.


These techniques are useful for detecting complex outliers in larger datasets.

# 8. Different Types of Outliers
1. Global Outliers
- A data point that is significantly different from the entire dataset.
Example: Salaries = 20k, 25k, 30k, 35k, 10 lakh.

2. Contextual Outliers
- A value that is unusual in a particular context.
Example: 35°C temperature is normal in summer but unusual in winter.

3. Collective Outliers
- A group of observations that becomes unusual when considered together.
Example: A sudden unusual pattern of transactions in a bank account.
