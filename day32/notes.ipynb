# 1. Encoding Numerical Features
Converting numerical features into a suitable numerical representation for Machine Learning algorithms.

Types:
- Continuous: Height, Weight, Salary.
- Discrete: Number of students, Cars.
 
# 2. Discretization
Discretization is the process of converting continuous numerical data into discrete intervals called Bins.

Why use it?
- Handle outliers.
- Simplify data.
- Identify patterns.

# 3. Types of Discretization
binning :-
i. unsupervised
ii. supervised (type - decision tree binning)
iii. custom binning

Mainly 3 types:(unsupervised)
1. Equal Width Binning(uniform)
2. Equal Frequency Binning(quantile)
3. KMeans Binning

# 4. Equal Width / Uniform Binning
Divides data into intervals having equal width.

Formula:

Width = Max-Min/Number of bin

Python: pd.cut()

# 5. Equal Frequency / Quantile Binning
Divides data into bins containing approximately an equal number of observations.

Python: pd.qcut()

# 6. KMeans Binning
Uses the KMeans clustering algorithm to group similar numerical values into bins.

Python: strategy='kmeans'

# 7. Encoding the Discretized Variable
Converts categorical bins into numerical values.

Types:
- Ordinal Encoding
- One-Hot Encoding

# 8. Example
Age → Age Groups:
- 0–18 = Child
- 19–35 = Young
- 36–60 = Adult
- 60+ = Senior

# 9. Custom / Domain Based Binning
Creates bins manually using domain knowledge or business requirements.

Example:
- Salary < 20K = Low
- 20K–50K = Medium
- Salary > 50K = High

# 10. Binarization
Converts numerical values into binary values (0 and 1) using a threshold.

Rule:
- Value > Threshold = 1
- Value ≤ Threshold = 0
Python: Binarizer()

# 11. Example (Binarization)
Suppose threshold = 50.
Marks	Output
20	0
40	0
50	0
60	1
80	1

Remember:
