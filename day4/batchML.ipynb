# 1. Batch vs Online Machine Learning

Machine Learning ke training approaches ke 2 main types hain:

A. Batch Learning

Batch Learning mein model ko poore available training dataset par ek saath train kiya jata hai.

Example: Ek company ke paas 1 lakh customers ka data hai. Model ko ek saath poore data par train karna.

B. Online Learning

Online Learning mein model new data se continuously ya small batches mein incrementally learn karta hai.

Example: Online shopping website par new customer transactions aati hain aur model unse gradually learn karta hai.

Batch Learning	Online Learning
Trains on the complete dataset	Learns from incoming data
Periodic training	Continuous/incremental training
Retraining required for new data	Can update with new data
Less adaptive	More adaptive

### 2. Batch / Offline Machine Learning

Definition: Batch Learning is a training approach in which the model is trained offline using the complete available dataset.

Working Process
Collect the complete dataset.
Train the model.
Test the model.
Deploy the trained model.
Collect new data.
Retrain the model periodically.

Example:

Suppose a bank has 1 lakh customer records.

First, model is trained using 1 lakh records.
Then the model is deployed.
After one month, 10,000 new customer records arrive.
Model is retrained using the updated dataset.

Important: Batch Learning is also called Offline Learning because training generally happens separately from its normal prediction process.

### 3. The Problem with Batch ML

The biggest problem with Batch ML is that it cannot learn incrementally from new data automatically.

Whenever new data arrives, the model generally needs to be retrained.

Example: YouTube Recommendation System

Suppose YouTube has trained a recommendation model using previous users' watching history.

Now:

New videos are uploaded.
Users develop new interests.
Watching trends change.

But the existing Batch ML model does not automatically learn these new patterns.

To learn them, the model needs retraining.

Main Problems
Model cannot immediately adapt to new patterns.
Retraining is required.
Updating the model can take time.
Frequent retraining can consume resources.

### 4. Disadvantages of Batch ML

There are 3 major disadvantages:

1. Lots of Data

Batch ML needs to process the complete training dataset together, which becomes difficult when data is very large.

Example: Training a model using millions of Facebook posts.

2. Hardware Limitation

Batch ML may require high RAM, CPU and GPU resources to process large datasets.

Example: A 50 GB dataset may be difficult to load into a computer having only 8 GB RAM.

3. Availability

Batch ML cannot automatically learn from newly arriving data. It needs periodic retraining to update its knowledge.

Example: A shopping website receives new customer data every minute, but its Batch ML model will not automatically learn from every new transaction.
