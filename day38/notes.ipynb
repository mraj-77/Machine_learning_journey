# 1. Missing Indicator
- Definition: Missing Indicator is a technique used to create a separate binary column to identify whether a value was missing or not.
- 1 → Value was missing.
- 0 → Value was present.
- It helps ML models identify patterns related to missing data.
Syntax:
from sklearn.impute import MissingIndicatormi = MissingIndicator()X_train_missing = mi.fit_transform(X_train)X_test_missing = mi.transform(X_test)


Important Points:
- Does not fill missing values.
- Creates an additional indicator feature.
- Helps preserve information about missingness.

# 2. Random Imputation
- Definition: Random Imputation is a technique used to replace missing values by randomly selecting values from the existing non-missing values of the same column.
Working:
1. Identify missing values.
2. Collect non-missing values.
3. Randomly select values from existing data.
4. Replace missing values with selected values.
Python Code:
import numpy as npimport pandas as pdX_train['Age_imputed'] = X_train['Age'].copy()random_sample = X_train['Age'].dropna().sample(    X_train['Age'].isnull().sum(),    random_state=42)random_sample.index = X_train[X_train['Age'].isnull()].indexX_train.loc[X_train['Age'].isnull(), 'Age_imputed'] = random_sample


Advantages:
- Preserves the original data distribution approximately.
- Maintains variability.

- Simple to implement.
Disadvantages:
- Results depend on random selection.
- Can introduce noise.
- Requires careful handling of train and test data.

# 3. Automatically Imputer (Automatic Imputation)
- Definition: Automatic Imputation means using an automated pipeline or transformer to handle missing values without manually applying imputation separately to every column.
Python Code:
from sklearn.impute import SimpleImputerfrom sklearn.compose import ColumnTransformertrf = ColumnTransformer([    ('imputer', SimpleImputer(strategy='mean'), ['Age']),    ('imputer2', SimpleImputer(strategy='median'), ['Fare'])], remainder='passthrough')X_train = trf.fit_transform(X_train)X_test = trf.transform(X_test)


Advantages:
- Reduces manual work.
- Handles multiple columns.
- Maintains consistent preprocessing.
- Useful in ML pipelines.

Disadvantages:
- Requires correct column selection.
- Wrong strategy can affect model performance.
- Needs proper train-test separation.

Quick Revision
Technique	Main Purpose
Missing Indicator	Identifies missing values (0/1)
Random Imputation	Fills missing values using random samples
Automatic Imputation	Automates missing-value handling
Mean Imputation	Fills using mean
Median Imputation	Fills using median
Mode Imputation	Fills using most frequent value
KNN Imputation	Fills using nearest neighbors


Important: Automatic imputation is a general approach, not one specific imputation algorithm. In Scikit-learn, SimpleImputer, KNNImputer, and ColumnTransformer can be combined in pipelines for automated preprocessing.
