# Types of Feature Scaling(Normalization)

Image mein 4 techniques di gayi hain:

### 1. Min-Max Scaling (Normalization)

Values ko generally 0 se 1 ke beech convert karta hai.

Formula:

$$ X_{scaled} = \frac{X-X_{min}}{X_{max}-X_{min}} $$

Example:

Min = 10
Max = 50
X = 30
$$ \frac{30-10}{50-10} = 0.5 $$

Python:

from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(df)

### 2. Mean Normalization

Data ko mean (average) ke around center karta hai.

Formula:

$$ X_{scaled} = \frac{X-X_{mean}}{X_{max}-X_{min}} $$

### 3. Max Absolute Scaling

Data ko maximum absolute value se divide karta hai.

Formula:

$$ X_{scaled} = \frac{X}{|X|_{max}} $$

Python:

from sklearn.preprocessing import MaxAbsScaler

scaler = MaxAbsScaler()
X_scaled = scaler.fit_transform(df)

### 4. Robust Scaling

Median aur Interquartile Range (IQR) ka use karta hai.

Formula:

$$ X_{scaled} = \frac{X-X_{median}}{IQR} $$

Python:

from sklearn.preprocessing import RobustScaler

scaler = RobustScaler()
X_scaled = scaler.fit_transform(df)
