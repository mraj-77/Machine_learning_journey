# Encoding Categorical Variables

Definition:
Encoding is the process of converting categorical (text) data into numerical data so that Machine Learning algorithms can understand it.

## types of data 
1.numericl
2.categorical

## 1. Types of categorical Data

Categorical data is mainly divided into two types:

A. Nominal Data

Categories have no specific order.
Example: Cat, Dog, Cow, Red, Blue.

B. Ordinal Data

Categories have a meaningful order or ranking.
Example: Good, Better, Best.

### 2. Types of Encoding

A. Label Encoding

Converts each category into a unique numerical value.
Mostly used for categorical labels (target variable).

Example:

Category	Value
Cat	        0
Dog	        1
Cow	        2

B. Ordinal Encoding

Converts ordered categories into numerical values.
Maintains the order of categories.(Input varible)

Example:

Category	Value
Bad	        0
Good	    1
Excellent	2

C. One-Hot Encoding

Creates separate columns for each category.
Assigns 1 or 0 depending on the category.
Commonly used for nominal data.

Example:

Animal	Cat	Dog	Cow
Cat	1	0	0
Dog	0	1	0
Cow	0	0	1

### 3. Feature Engineering

Feature Engineering is the process of creating, transforming or selecting features to improve Machine Learning model performance.

Feature Transformation: Changing existing features into a suitable format for the model.

Example: Converting categorical data into numerical data using Encoding.

Important Points
Nominal → One-Hot Encoding (generally preferred).
Ordinal → Ordinal Encoding.
Target Labels → Label Encoding (commonly used).
Encoding helps ML algorithms process categorical data.

Remember:

Nominal = No order.
Ordinal = Ordered categories.
Label = Assign numbers.
One-Hot = Create separate columns.
