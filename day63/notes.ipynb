# Ridge Regression (L2 Regularization)

### 1. What is Regularization?
- Technique to prevent **Overfitting**.
- Adds a penalty to the Loss Function.
- Controls large model coefficients (slopes).

### 2. Normal Linear Regression

Without Regularization:

\[
Loss = \sum_{i=1}^{n}(y_i-\hat{y}_i)^2
\]

### 3. Ridge Regression (L2)

Adds a penalty to Linear Regression Loss:

\[
Loss = \sum_{i=1}^{n}(y_i-\hat{y}_i)^2 + \lambda\sum_{j=1}^{p}m_j^2
\]

- **λ (Lambda):** Regularization strength.
- **m:** Model coefficients (slopes).
- **m²:** Square of coefficients.
- Reduces large coefficients but generally doesn't make them zero.

### 4. Lambda (λ)

| Lambda | Effect |
|---|---|
| λ = 0 | Linear Regression |
| Small λ | Less Penalty |
| Large λ | More Penalty |
| Very Large λ | Coefficients move towards zero |

**Remember:**
- Low λ → Overfitting Risk.
- High λ → Underfitting Risk.

### 5. Types of Regularization

| Algorithm | Regularization | Penalty |
|---|---|---|
| Ridge | L2 | λ × Σ(m²) |
| Lasso | L1 | λ × Σ\|m\| |
| Elastic Net | L1 + L2 | Both |

### 6. Important Points
- Ridge → Reduces coefficients.
- Lasso → Can make coefficients exactly zero.
- Elastic Net → Combination of Ridge + Lasso.
- Regularization → Helps prevent Overfitting.
- Feature Scaling → Important before Regularization.
