# Types of Machine Learning

# 1. Supervised Learning

Model learns from labeled data (input with correct output).

Regression: Predicts continuous numerical values. Example: House price prediction.
Classification: Predicts categories or classes. Example: Spam or Not Spam.

### Types of Supervised Learning
A. Classification

Classification is used to predict categorical output (Class/Category).
Example: Student Placement Prediction.

Real-life examples:

Spam / Not Spam
Pass / Fail
Loan Approved / Rejected
Disease Positive / Negative

# 2. Unsupervised Learning

Model learns from unlabeled data and discovers hidden patterns.

Clustering: Groups similar data points together. Example: Grouping customers based on shopping habits.
Dimensionality Reduction: Reduces the number of features while preserving important information. Example: Reducing 100 features to 10.
Anomaly Detection: Identifies unusual or abnormal data. Example: Detecting fraudulent bank transactions.
Association: Finds relationships between different items. Example: Customers who buy bread also buy butter.

### Unsupervised Learning

Definition: Unsupervised Learning is a type of Machine Learning in which a model learns from unlabeled data and discovers hidden patterns without predefined outputs.

Unsupervised Learning has 4 main tasks:

## 1. Clustering

Clustering is used to group similar data points into different clusters.

Example: Student Grouping

Suppose we have data of 5,000 students:

IQ	CGPA
70	7.0
60	6.0
100	9.0
55	5.5
110	9.5

The model automatically groups students based on similarities in IQ and CGPA.

Real-life examples:

Customer segmentation
Student grouping
Grouping similar products
Social media audience grouping

Algorithms: K-Means, Hierarchical Clustering, DBSCAN.

## 2. Dimensionality Reduction

Dimensionality Reduction is used to reduce the number of features (columns) while preserving important information.

Example:

Suppose a dataset has 100 features. We reduce them to 10 important features.

100 Features → 10 Features

Why use it?

Reduces complexity.
Saves memory and computation time.
Helps visualize high-dimensional data.
Removes redundant information.
Main techniques:
PCA (Principal Component Analysis): Converts original features into fewer new features while retaining maximum possible variance.
Feature Selection: Selects the most important existing features.
Feature Extraction: Creates new features from existing features.

## 3. Anomaly Detection

Anomaly Detection is used to identify unusual or abnormal data points that are different from normal data.

Example:

A person usually spends ₹500–₹2,000 daily, but suddenly a transaction of ₹2,00,000 occurs.

The system may identify this transaction as an anomaly.

Real-life examples:

Credit card fraud detection
Network security
Unusual banking transactions
Machine failure detection

Algorithms: Isolation Forest, One-Class SVM.

## 4. Association Rule Learning

Association Rule Learning is used to discover relationships between different items in a dataset.

Example: Market Basket Analysis

Suppose customers frequently purchase:

Bread + Butter
Milk + Bread
Bread + Butter + Jam

The model discovers that customers who buy bread may also buy butter.

Real-life examples:

Amazon product recommendations
Grocery store product placement
Netflix recommendations (though Netflix uses several ML approaches)
Shopping recommendations

Algorithms: Apriori, FP-Growth.


# 3. Semi-Supervised Learning

Model learns using a small amount of labeled data and a large amount of unlabeled data.

Example: Image recognition using 100 labeled images and 10,000 unlabeled images.

# 4. Reinforcement Learning

Model learns by taking actions and receiving rewards or penalties.

Example: Training an AI to play chess or a robot to walk.
