# Gradient Descent – Short Notes (Easy Hinglish)

# 1. What is Gradient Descent?
Gradient Descent ek optimization algorithm hai jo Loss Function ko minimum karne ke liye best parameters find karta hai.

Simple meaning: Model ki error ko kam karna.

# 2. Intuition
- Sabse pehle random value of b select karte hain.
- Loss calculate karte hain.
- Slope (gradient) check karte hain.
- Slope ke according b ko update karte hain.
- Ye process repeat hota hai jab tak minimum loss nahi milta.

# 3. Slope (Gradient)
Slope	Direction
Positive (+)	Left direction
Negative (-)	Right direction
Zero (0)	Minimum point (for a convex loss function)


# 4. Mathematical Formula
Parameter Update:
\[
b_{new} = b_{old} - \alpha \times slope
\]
Where:
- b_old = Current parameter
- b_new = Updated parameter
- α (Alpha) = Learning Rate
- Slope = Gradient / Derivative

# 5. Example
Suppose:
\[
L(b) = b^2
\]
Derivative:
\[
\frac{dL}{db} = 2b
\]
Agar b = -10:
\[
Slope = 2(-10) = -20
\]
Slope negative hai, toh parameter right direction mein move karega.

# 6. Learning Rate
Learning Rate decide karta hai ki ek step mein kitna move karna hai.
Learning Rate	Problem
Very Small	Training slow
Very Large	Minimum point miss/overshoot ho sakta hai
Appropriate	Faster convergence


# Interview One-Liner
Gradient Descent is an optimization algorithm that minimizes the loss function by repeatedly updating model parameters in the opposite direction of the gradient.
Remember This 🧠
- Gradient = Slope
- Loss = Error
- Learning Rate = Step Size
- Goal = Minimum Loss
- Direction = Opposite of Gradient
Golden Formula:
\[
\boxed{b_{new}=b_{old}-\alpha \frac{dL}{db}}
\]
