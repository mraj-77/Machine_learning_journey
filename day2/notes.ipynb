# AI vs ML vs DL

# 1. Artificial Intelligence (AI)

Definition: AI is a technology that enables machines to perform tasks that normally require human intelligence.

Examples: ChatGPT, Robots, Google Assistant.


# 2. Machine Learning (ML)

Definition: ML is a subset of AI that enables computers to learn from data and make predictions without being explicitly programmed for every task.

Examples: Spam Detection, House Price Prediction, YouTube Recommendations.


# 3. Deep Learning (DL)

Definition: DL is a subset of ML that uses multi-layered neural networks to learn complex patterns from large amounts of data.

Examples: Face Recognition, Voice Recognition, Self-driving Cars.
