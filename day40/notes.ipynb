# 1. What is Iterative Imputer? or MICE
Definition: Iterative Imputer is a missing value imputation technique that predicts missing values using other available features (columns) in the dataset.

It uses relationships between multiple columns instead of simply calculating mean or median.

MICE stands for: Multivariate Imputation by Chained Equations.
- Multivariate → Multiple features/columns.
- Imputation → Filling missing values.
- Chained Equations → Predicting missing values one column at a time, repeatedly.

# 2. Assumptions of MICE
MICE is commonly used under the following missing-data mechanisms:
Type	Full Form	                                Meaning
MCAR	Missing Completely At Random	    Missing values have no relationship with any variable.
MAR	    Missing At Random	                Missingness depends on other observed variables.
MNAR	Missing Not At Random	            Missingness depends on the missing value itself or other unobserved information.


Important: MICE is generally most defensible under the MAR assumption. It can also be applied in other situations, but MNAR may require additional modelling assumptions.

# 3. How Does Iterative Imputer Work?
Let's understand with an example.

Suppose we have this dataset:
Age	Experience	Salary
20	1	        20000
25	3	        30000
NaN	5	        40000
30	NaN	        50000
35	7	        NaN


There are missing values in three columns.
### Step 1: Initial Imputation
First, Iterative Imputer temporarily fills missing values using a simple strategy (by default, mean).
Age	Experience	    Salary
20	    1	        20000
25	    3	        30000
27.5	5	        35000
30	    4	        50000
35	    7	        35000


Now there are no missing values.

### Step 2: Select First Column (Age)
- Age becomes our target variable (Y).
- Experience and Salary become independent variables (X).
- A regression model is trained using rows where Age is available.
- The model predicts the missing Age.
Example:
Experience + Salary → Model → Predict Age

Suppose predicted Age = 28.

### Step 3: Select Second Column (Experience)
Now Experience becomes the target variable.
- Age and Salary are used as input features.
- The model learns their relationship with Experience.
- It predicts the missing Experience.
Age + Salary → Model → Predict Experience

### Step 4: Select Third Column (Salary)
Now Salary becomes the target variable.
- Age and Experience become input features.
- The model predicts missing Salary.
Age + Experience → Model → Predict Salary

### Step 5: Repeat the Process
This is the most important step!

The algorithm repeats the entire process:
1. Predict missing Age.
2. Predict missing Experience.
3. Predict missing Salary.
4. Again update Age.
5. Again update Experience.
6. Again update Salary.
This process continues until:
- Values become stable (convergence), or
- Maximum iterations are reached.
Remember: Iterative Imputer works column by column and repeats the process to improve missing-value predictions.

# 5. Advantages
1. Accurate: Uses relationships between multiple features.
2. Multivariate: Considers multiple columns simultaneously.
3. Flexible: Can use different ML models for prediction.
4. Better information: Can preserve relationships between variables better than simple mean imputation.

# 6. Disadvantages
1. Slow: Repeatedly trains models for different columns.
2. Memory: Can consume more memory.
3. Complex: More difficult to understand and implement.
4. Assumptions: Results depend on the quality of the predictive models and missing-data assumptions.
7. Important Difference: Simple vs KNN vs Iterative Imputer

Simple Imputer	KNN Imputer	Iterative Imputer
Mean/Median/Mode	Nearest neighbors	Predictive models
Simple	Distance-based	Model-based
Fast	Comparatively slow	Comparatively complex
Column-wise statistics	Similar rows	Relationships between features


8. Important Interview Questions
Q1. What is MICE?

MICE stands for Multivariate Imputation by Chained Equations. It fills missing values by predicting each incomplete feature using other available features.

Q2. What is the main advantage of Iterative Imputer?

It considers relationships between multiple features while filling missing values.

Q3. Why is it called Iterative Imputer?

Because it repeatedly predicts and updates missing values across different columns.

Q4. Which algorithm does Iterative Imputer use by default in Scikit-learn?

Bayesian Ridge Regression.

One-line trick to remember:
- Mean Imputer → Average se fill.
- Random Imputer → Random values se fill.
- KNN Imputer → Nearest neighbors se fill.
- Iterative Imputer → ML model se predict karke fill.
