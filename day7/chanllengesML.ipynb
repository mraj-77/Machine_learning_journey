# 1. Data Collection

ML ke liye pehle data chahiye. Data lene ke 2 main tarike:

API: Website/company ek official “darwaza” deti hai jahan se aap data le sakte ho (jaise Twitter API, weather API).
Web Scraping: Jab API nahi hoti, to code likhkar website se data nikalte hain.

# 2. Insufficient Data / Labelled Data

Data kam hai, ya data par label (sahi answer) nahi laga.
Example: 10 photos se aap “kutta vs billi” nahi sikha sakte. Aur har photo par likha hona chahiye ki ye kutta hai ya billi. Labelling mehngi aur time-consuming hoti hai.

# 3. Non-Representative Data

Training data real duniya ko represent nahi karta.
Example: Aap sirf Delhi ke logon ka data lekar poore India ke liye model banao, to result galat aayega.

# 4. Poor Quality Data

Data mein galtiyan, missing values, noise, duplicates hain.
Example: Kisi ki age 500 likhi hai, ya kuch columns khali hain. Rule: “Garbage In, Garbage Out” (kachra daloge to kachra hi niklega).

# 5. Irrelevant Features

Data mein aise columns hain jo prediction mein kaam hi nahi aate.
Example: Ghar ka price predict karna hai, aur data mein “malik ka favourite color” bhi hai. Iska koi fayda nahi. Solution: Feature selection (sirf kaam ke columns rakho).

# 6. Overfitting

Model training data ko ratta maar leta hai, sirf wahi par accha chalta hai, naye data par fail.
Example: Student ne sirf purane paper ke answers ratte, naya question aate hi fail.

# 7. Underfitting

Model bahut simple hai, data ka pattern hi nahi samajh paaya. Training par bhi kharab, test par bhi kharab.
Example: Student ne padhai hi nahi ki, to purane aur naye dono questions mein fail.

Yaad rakho: Overfitting = zyada ratta, Underfitting = kuch seekha hi nahi.

# 8. Software Integration

Model bana liya, par use asli software/app/website mein jodna mushkil hota hai.
Example: Aapne Jupyter Notebook mein accha model banaya, par ab use mobile app ya website mein chalana hai. Iske liye APIs, servers aur engineering skills chahiye.

# 9. Offline Learning / Deployment

Offline learning: Model ek baar train hota hai, phir deploy ho jata hai. Naya data aane par use dobara poora train karna padta hai.
Problem: Duniya badalti rehti hai (trends, user behavior), to purana model dheere-dheere purana ho jata hai. Isliye model ko regular update karna padta hai. Iska alternative hai Online Learning, jisme model naye data se lagataar seekhta rehta hai.

# 10. Cost Involved

ML project mein paisa aur time dono lagte hain:

Data collection aur labelling ka kharcha
Powerful computers/GPU/cloud ka kharcha
Skilled logon (data scientists, engineers) ki salary
Deployment aur maintenance ka kharcha
