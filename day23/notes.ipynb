# Feature Engineering (Complete Notes)

Feature Engineering = raw data se useful features banana/nikalna, domain knowledge (us field ki samajh) use karke, taaki ML model ki performance behtar ho.

Feature = data ka ek column jisse model seekhta hai
Domain knowledge = us field ki jaankari (jaise ghar ke price ke liye real estate ki samajh)

##### Example:

Raw data: Date of Birth = 15-08-2000
Feature banaya: Age = 25

Model ko raw date samajh nahi aati, lekin age jaisa feature se pattern aasani se mil jata hai.

Feature Engineering ke 4 Main Parts
#	Part	                    Ek Line Mein
1	Feature Transformation	Data ko model ke layak banana
2	Feature Construction	Purane features se naye features banana
3	Feature Selection	Sirf kaam ke features chunna
4	Feature Extraction	Bahut saare features ko kam features mein badalna

### 1. Feature Transformation

Data ko model ke samajhne layak form mein badalna. Isme 4 sub-parts hain:

##### 1.1 Missing Value Imputation

Data mein jo values khali (missing) hain, unhe bharna.

Mean/Median: Number columns ke liye (age khali hai to average age bhar do)
Mode: Category columns ke liye (sabse common value)
Example: Age column mein 5 log ki age missing hai, to unki jagah average age 30 bhar do.

##### 1.2 Handling Categorical Features

Model numbers samajhta hai, text nahi. To text categories ko numbers mein badalna padta hai.

Label Encoding: Small=0, Medium=1, Large=2
One-Hot Encoding: Har category ka alag column
City: Delhi, Mumbai  →  City_Delhi (1/0), City_Mumbai (1/0)

##### 1.3 Outlier Detection

Outlier = aisi value jo baaki data se bahut alag ho.

Example: Sab logon ki salary 30k-80k hai, lekin ek ki 50 crore hai. Ye model ko confuse kar sakta hai.
Tarike: Box plot, Z-score, IQR method
Kya karein: Hata do ya cap kar do

##### 1.4 Feature Scaling

Alag-alag range wali values ko same scale par laana.

Example: Age (0-100) aur Salary (0-1,00,000). Bina scaling ke model salary ko zyada important samajh lega.
Min-Max Scaling: Values ko 0 se 1 ke beech laata hai
Standardization: Mean 0 aur Std Dev 1 banata hai

### 2. Feature Construction

Maujooda features se naye features banana.

Example: Height aur Weight se BMI = Weight / Height²
Example: Date se Day, Month, Is_Weekend
Example: Total_Price = Quantity × Price

Isme domain knowledge sabse zyada kaam aata hai.

### 3. Feature Selection

Faltu aur kam kaam ke features hata dena, sirf important features rakhna.

Example: Ghar ke price ke liye “malik ka favourite color” faltu hai.

Fayde:

Model fast train hota hai
Overfitting kam hoti hai
Accuracy badhti hai

Tarike: Correlation check, Forward/Backward Selection, Feature Importance

### 4. Feature Extraction

Bahut saare features ko kam features mein convert karna, bina zyada information khoye. Naye features original se bane hote hain, lekin alag hote hain.

Example: 1000 columns ko PCA se 50 columns mein badalna
Example: Image se edges aur shapes nikalna, text se keywords nikalna
Construction vs Extraction vs Selection
Point	Construction	Selection	Extraction
Kya karta hai	Naye features banata hai	Purane mein se chunta hai	Features ko compress karta hai
Features ki ginti	Badhti hai	Kam hoti hai	Kam hoti hai
Example	Height, Weight → BMI	Faltu columns hatana	PCA

##### Kyun Zaroori Hai?
Model ki accuracy badhti hai
Model fast train hota hai
Overfitting kam hoti hai
Data model ke samajhne layak banta hai

Famous line: “Better features beat better algorithms.”

Yaad Rakhne Ka Trick

TCSE = Transformation, Construction, Selection, Extraction

Saaf karo → Naye banao → Chuno → Compress karo → Model train karo

##### Python Example

import pandas as pd
from sklearn.preprocessing import MinMaxScaler

df = pd.DataFrame({
    'height_m': [1.7, 1.8, 1.6],
    'weight_kg': [70, 90, None],
    'city': ['Delhi', 'Mumbai', 'Delhi'],
    'dob': ['2000-08-15', '1995-03-10', '1998-01-01']
})

# 1. Transformation: Missing value imputation
df['weight_kg'] = df['weight_kg'].fillna(df['weight_kg'].mean())

# 1. Transformation: Categorical encoding (One-Hot)
df = pd.get_dummies(df, columns=['city'])

# 2. Construction: Naye features
df['bmi'] = df['weight_kg'] / (df['height_m'] ** 2)
df['age'] = 2025 - pd.to_datetime(df['dob']).dt.year

# 3. Selection: Faltu column hatana
df = df.drop(columns=['dob'])

# 1. Transformation: Feature Scaling
scaler = MinMaxScaler()
df[['height_m', 'weight_kg']] = scaler.fit_transform(df[['height_m', 'weight_kg']])

print(df)
