# 1. Handling Missing Data
Definition: Handling Missing Data is the process of identifying and treating missing values (NaN) in a dataset before applying Machine Learning algorithms.

Missing values occur when some information is not available in a dataset.

Example:
Name	Age	Salary
Ajay	22	25000
Rahul	NaN	30000
Aman	25	NaN
Rohit	28	35000


Main techniques:
1. Removing Missing Values
2. Imputation (Filling Missing Values)

# 2. Complete Case Analysis (CCA)
Definition: Complete Case Analysis is a technique in which rows containing missing values are completely removed from the dataset.

It is also called Listwise Deletion.

Example:

Before CCA:
Name	Age	Salary
Ajay	22	25000
Rahul	NaN	30000
Aman	25	NaN
Rohit	28	35000


After CCA:
Name	Age	Salary
Ajay	22	25000
Rohit	28	35000


Python:
df.dropna()


# 3. Assumptions for CCA
CCA works best when missing values are distributed randomly.

Main assumption: MCAR (Missing Completely At Random)

MCAR means missing values have no relationship with any variable in the dataset.

Example: Some customer records are missing because of a random technical error.

Other missing data mechanisms:
- MCAR: Missing values occur completely randomly.
- MAR: Missing values depend on other observed variables.
- MNAR: Missing values depend on the missing value itself.

# 4. Advantages and Disadvantages of CCA
Advantages:
1. Easy to understand and implement.
2. No need to calculate missing values.
3. Maintains original values of remaining observations.
4. Simple preprocessing technique.

Disadvantages:
1. Loss of valuable data.
2. Reduces dataset size.
3. Can introduce bias if data is not MCAR.
4. May reduce model performance.

# 5. When to Use CCA?
CCA is suitable when:
1. Missing values are very low (generally 5% or less, as a rough guideline).
2. Data is MCAR.
3. Dataset contains sufficient observations.
4. Removing rows does not significantly affect data distribution.
Avoid CCA when a large number of rows contain missing values.

# 6. Imputation
Definition: Imputation is the process of replacing missing values with estimated or calculated values instead of deleting rows.

Imputation has two main types:
A. Simple Imputation (Univariate)
Univariate imputation uses a single feature/column to calculate replacement values.

It is divided into:

### 1. Numerical Features
- Mean Imputation: Replace missing values with the average.
- Median Imputation: Replace missing values with the middle value.
- Random Imputation: Replace missing values with randomly selected existing values.
- End of Distribution: Replace missing values with an extreme value at the end of the distribution.

### 2. Categorical Features
- Mode Imputation: Replace missing values with the most frequent category.
- Missing Category: Replace missing values with a new category called 'Missing'.
B. Multivariate Imputation
Multivariate imputation uses relationships between multiple features to estimate missing values.

### 1. KNN Imputation

K-Nearest Neighbors uses similar data points (nearest neighbors) to estimate missing values.

### 2. Iterative Imputation (MICE)

Iterative Imputation predicts missing values using other features through repeated modeling.

MICE stands for Multiple Imputation by Chained Equations.
