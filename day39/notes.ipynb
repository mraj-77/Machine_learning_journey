# 1. Definition: KNN Imputer is a technique used to handle missing values (NaN) by finding K nearest neighbors and using their values to fill missing data.

2. Working:
1. Finds missing values.
2. Calculates distance between data points.(using nan_euclidean)
3. Finds K nearest neighbors.
4. Uses their average to fill missing values.
3. Syntax:
from sklearn.impute import KNNImputerimputer = KNNImputer(n_neighbors=5)X_train = imputer.fit_transform(X_train)X_test = imputer.transform(X_test)


### 4. Parameters:
- n_neighbors: Number of nearest neighbors (default = 5).
- weights: uniform or distance.
- metric: Distance calculation method.

### 5. Advantages:
- Considers relationships between features.
- Can provide better imputation than mean/median.
- Uses similar data points.

### 6. Disadvantages:
- Computationally expensive.
- Sensitive to feature scaling.
- Performance depends on K value.
- Not directly suitable for categorical data.

### 7. Important Points:
- Works with numerical data.
- Feature scaling is important.
- Fit on training data only.
- Apply transform on test data.

### 8. Mean vs KNN Imputer:
- Mean Imputer → Fills missing values using column mean.
- KNN Imputer → Fills missing values using nearest neighbors.
Remember: KNN Imputer = Missing Values + Distance + Nearest Neighbors.
