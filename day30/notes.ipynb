# 📘 ML Notes: Mathematical Transformations & Function Transformer

# 1. Mathematical Transformations
What is Mathematical Transformation?
Jab hum kisi numerical feature par mathematical function apply karke uski values ya distribution change karte hain, use Mathematical Transformation kehte hain.

# Why do we use it?
1. Skewness reduce karne ke liye.
2. Data ko Normal Distribution ke closer lane ke liye.
3. Large values ka effect reduce karne ke liye.
4. Kuch ML algorithms ki performance improve karne ke liye.

# Types of Mathematical Transformations
### 1. Log Transformation

Formula:

f(x)=\log(x)

Large values ko compress karta hai aur right skewness reduce karne mein helpful hai.

### 2. Reciprocal Transformation

Formula:

f(x)=\frac{1}{x}

Har value ka reciprocal calculate karta hai.

### 3. Power Transformation

Ismein feature ki power change karte hain.

Square:

f(x)=x^2

Square Root:

f(x)=\sqrt{x}

### 4. Custom Transformation

Apni requirement ke according mathematical function apply karna.

Example:

f(x)=x^2+2x+1

# 2. Function Transformer
What is FunctionTransformer?
Scikit-learn ka FunctionTransformer ek tool hai jo kisi feature par custom mathematical function apply karne mein help karta hai.

Example: Log Transformation
import numpy as npfrom sklearn.preprocessing import FunctionTransformerft = FunctionTransformer(np.log1p)X = [[1], [10], [100], [1000]]X_transformed = ft.fit_transform(X)print(X_transformed)


Different Transformations using FunctionTransformer
Log Transformation
ft = FunctionTransformer(np.log1p)


Square Transformation
ft = FunctionTransformer(np.square)


Square Root Transformation
ft = FunctionTransformer(np.sqrt)


Reciprocal Transformation
ft = FunctionTransformer(np.reciprocal)


Custom Transformation
def custom(x):    return x**2 + 2*x + 1ft = FunctionTransformer(custom)


Remember: Reciprocal mein zero allowed nahi hai. Square root ke liye non-negative values honi chahiye.

# 3. How to Find if Data is Normal?
What is Normal Distribution?
Normal Distribution ek bell-shaped, symmetric distribution hoti hai.







Characteristics:
- Bell-shaped curve.
- Mean = Median = Mode.
- Data approximately symmetric hota hai.
- Most values mean ke aas-paas hoti hain.

## How to check Normal Distribution?
1. Histogram
import matplotlib.pyplot as pltplt.hist(df['Age'], bins=30)plt.show()


Agar histogram approximately bell-shaped hai, toh data normal ho sakta hai.

2. Skewness
df['Age'].skew()


Skewness	Meaning
0	Symmetric distribution
Positive	Right Skewed
Negative	Left Skewed


Generally, skewness zero ke close hone par distribution more symmetric hoti hai.

3. Statistical Test (Shapiro-Wilk Test)
from scipy.stats import shapirostat, p = shapiro(df['Age'].dropna())print(p)


Interpretation:
- p > 0.05 → Normality reject karne ka sufficient evidence nahi hai.
- p <= 0.05 → Data normal distribution follow nahi karta, aisa evidence milta hai.
Note: Large datasets mein test bahut sensitive ho sakta hai.

4. QQ Plot (Quantile-Quantile Plot)
What is QQ Plot?
QQ Plot ka use check karne ke liye hota hai ki given data Normal Distribution follow karta hai ya nahi.


Interpretation
- Points diagonal straight line ke close → Data approximately normal.
- Points line se bahar → Data normality se deviate kar raha hai.
- Curved pattern → Skewness ho sakti hai.
- Extreme ends par deviation → Tail behaviour ya outliers ho sakte hain.

import scipy.stats as statsimport matplotlib.pyplot as pltstats.probplot(df['Age'].dropna(), dist='norm', plot=plt)plt.show()


5. Log Transform
What is Log Transformation?
Log transformation mein feature ki har value ka logarithm calculate karte hain.

Formula:

f(x)=\log(x)

Example:
Original Value	Log Value (approx.)
1	            0
10	            2.30
100	            4.61
1000	        6.91


Notice karo: Original values ka difference bahut large hai, lekin log transformation ke baad values compress ho gayi hain.
Python Code
import numpy as npdf['log_price'] = np.log1p(df['Price'])


np.log1p(x) ka meaning hai:

\log(1+x)

Advantages:
- Right skewness reduce kar sakta hai.
- Large values compress karta hai.
- Income, population aur house prices mein useful hai.

Important:
- np.log(x) ke liye x positive hona chahiye.
- np.log1p(x) ke liye x > -1 hona chahiye.

#### 6. Other Transforms
Is section mein important transformations hain:

A. Reciprocal Transformation
Formula:

f(x)=\frac{1}{x}

df['reciprocal'] = 1 / df['Age']


- Large values ko small values mein convert karta hai.
- Zero allowed nahi hai.

B. Square Transformation
Formula:

f(x)=x^2

df['square'] = np.square(df['Age'])


- Distribution ki shape change karta hai.
- Kuch left-skewed distributions ko improve kar sakta hai.

C. Square Root Transformation
Formula:

f(x)=\sqrt{x}

df['sqrt'] = np.sqrt(df['Age'])


- Right skewness reduce karne mein useful hai.
- Negative values allowed nahi hain.

D. Box-Cox Transformation
Scikit-learn ka PowerTransformer Box-Cox method provide karta hai.
from sklearn.preprocessing import PowerTransformerpt = PowerTransformer(method='box-cox')X_transformed = pt.fit_transform(X)


Important: Box-Cox mein strictly positive values required hain.

E. Yeo-Johnson Transformation
pt = PowerTransformer(method='yeo-johnson')X_transformed = pt.fit_transform(X)


Advantages:
- Positive values handle karta hai.
- Zero values handle karta hai.
- Negative values bhi handle karta hai.
Box-Cox vs Yeo-Johnson
Box-Cox	Yeo-Johnson
Positive values only	Positive, zero, negative
Zero allowed nahi	Zero allowed
Power transformation	Power transformation


F. Quantile Transformer
Feature ki distribution ko selected distribution mein convert karta hai.

Two options:
1. Uniform Distribution
2. Normal Distribution
from sklearn.preprocessing import QuantileTransformerqt = QuantileTransformer(    output_distribution='normal',    random_state=42)X_transformed = qt.fit_transform(X)

Important Interview Point
Kya har ML algorithm ko Normal Distribution chahiye?

Nahi!
- Linear Regression: Transformation helpful ho sakti hai.
- Logistic Regression: Helpful ho sakti hai.
- Decision Tree: Generally normal distribution required nahi.
- Random Forest: Generally normal distribution required nahi.
Golden Rule: Training data par fit_transform() aur testing data par sirf transform() use karo.
