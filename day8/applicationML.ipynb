# 1. Retail: Amazon / Big Bazaar

Image mein supermarket dikha rahe hain, kyunki retail mein ML bahut use hota hai.

ML kya karta hai:

Recommendation System: “Aapne ye khareeda, to ye bhi pasand aayega” (Amazon pe “Customers also bought”)
Market Basket Analysis: Kaun si cheezein saath mein kharidi jaati hain. Jaise bread ke saath butter aur jam. Store mein inhe paas-paas rakhte hain.
Demand Forecasting: Kaunsa saaman kitna bikega, ye pehle se predict karna, taaki stock kam ya zyada na ho.
Price Optimization: Sahi price decide karna.

# 2. Banking and Finance

ML kya karta hai:

Fraud Detection: Agar aapke card se achanak ajeeb jagah se transaction ho, to system turant alert karta hai.
Credit Scoring / Loan Approval: Aapko loan dena safe hai ya nahi, ye predict karna.
Stock Market Prediction: Share prices ka trend samajhna.
Customer Support Chatbots: Bank ke automatic assistants.

# 3. Transport: OLA

ML kya karta hai:

Fare Prediction: Distance, time aur traffic dekhkar kiraya decide karna.
Surge Pricing: Demand zyada ho (barish, office time), to price badh jata hai.
ETA Prediction: Driver kitne minute mein aayega.
Driver-Rider Matching: Sabse nazdeeki driver ko aapse jodna.
Route Optimization: Sabse tez raasta dhundhna.

# 4. Manufacturing: Tesla

ML kya karta hai:

Self-Driving (Autopilot): Camera aur sensors se car road, signal aur logon ko pehchanti hai.
Predictive Maintenance: Machine kharab hone se pehle hi pata chal jata hai ki repair chahiye.
Quality Control: Factory mein camera se defective parts pakadna.
Robots: Assembly line par automation.

# 5. Consumer Internet: Twitter

ML kya karta hai:

Feed Ranking: Aapko kaunse tweets pehle dikhane hain.
Content Moderation: Hate speech, spam aur fake accounts pakadna.
Sentiment Analysis: Log kisi topic par positive hain ya negative.
Trending Topics: Kya viral ho raha hai.
Who to Follow: Naye log suggest karna.
