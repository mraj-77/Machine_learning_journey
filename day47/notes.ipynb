# PCA (Principal Component Analysis)
1. Introduction to PCA
Full Form: Principal Component Analysis.

Definition: PCA is a dimensionality reduction technique used to reduce the number of features in a dataset while preserving as much important information (variance) as possible.

In simple words:
PCA converts a large number of features into a smaller number of new features called Principal Components.

Example:

Suppose a dataset has 100 features.
100 Features
     |
     v
    PCA
     |
     v
10 Principal Components

We reduce 100 dimensions to 10 dimensions while trying to preserve maximum information.

Important Points:
- PCA is an unsupervised learning technique.
- It is used for dimensionality reduction.
- It helps reduce computational complexity.
- It can help reduce noise and overfitting.
- PCA creates new features instead of selecting existing features.

Important: PCA does not randomly remove columns. It finds the best directions to represent the data with minimum information loss.

# 3. Why Is Variance Important?
Definition: Variance measures how much data points are spread out from their mean.

In simple words:
- High variance → Data points are more spread out.
- Low variance → Data points are closer together.
Example
Suppose we have two features:
Student	Study Hours	Sleep Hours
A	1	7
B	2	7
C	5	7
D	8	7
E	10	7


Here:
- Study Hours has high variance.
- Sleep Hours has zero variance.
Why?

Because Study Hours values are spread out, while Sleep Hours has the same value for everyone.
Why does PCA care about variance?
PCA assumes that directions with higher variance contain more useful information about the structure of the data.

Therefore:
- PCA identifies directions with maximum variance.
- It gives priority to high-variance directions.
- It tries to preserve maximum variance using fewer dimensions.
Remember: PCA tries to retain maximum variance with minimum dimensions.

Note: High variance does not always mean useful information. PCA can also preserve noise if the noise has high variance.

# 4. Problem Formulation
Suppose we have a dataset with two features:
- X1 = Height
- X2 = Weight
We want to reduce these two features into one feature.

Original data:
X1 = Height
X2 = Weight

PCA finds a new direction along which the data has maximum variance.

This new direction is called the First Principal Component (PC1).
Original Features
      X1, X2
         |
         v
        PCA
         |
         v
        PC1

Main Objective of PCA
PCA tries to:
1. Find the direction of maximum variance.
2. Project data onto that direction.
3. Preserve maximum information.
4. Reduce the number of dimensions.
Mathematically:
PC1 = w_1X_1+w_2X_2

Where:
- X1, X2 = Original features.
- w1, w2 = Weights (loadings).
- PC1 = First Principal Component.

# 5. Covariance and Covariance Matrix
A. What is Covariance?
Definition: Covariance measures how two variables change together.

There are three possibilities:
Covariance	Meaning
Positive	Both variables increase or decrease together.
Negative	One variable increases while the other decreases.
Zero	No linear relationship is indicated.


Example:

Height and Weight:

Generally, when height increases, weight may also increase.

So they may have positive covariance.
B. What is Covariance Matrix?
A covariance matrix shows the covariance between every pair of features.

Suppose we have two features X1 and X2.

Covariance Matrix:

C =
\begin{bmatrix}
Var(X_1) & Cov(X_1,X_2)\\
Cov(X_2,X_1) & Var(X_2)
\end{bmatrix}

Where:
- Diagonal values represent variance.
- Off-diagonal values represent covariance.
For example:

C =
\begin{bmatrix}
4 & 2\\
2 & 9
\end{bmatrix}

Here:
- Variance of X1 = 4.
- Variance of X2 = 9.
- Covariance between X1 and X2 = 2.
Important: The covariance matrix helps PCA understand the spread and relationships between features.

# 6. Linear Transformation and Eigenvectors
This is one of the most important mathematical concepts in PCA.
A. Linear Transformation
A linear transformation changes the representation of data using mathematical operations such as rotation, scaling or projection.

In PCA, we transform the original features into new directions.
Original Data (X1, X2)
          |
          v
  Linear Transformation
          |
          v
 New Features (PC1, PC2)

B. Eigenvectors
Definition: Eigenvectors represent the directions along which data has maximum variance or other principal directions of the covariance matrix.

In PCA:
- Eigenvectors give the directions of principal components.
- PC1 is the direction with maximum variance.
- PC2 is the direction with the second-highest variance.
C. Eigenvalues
Definition: Eigenvalues represent the amount of variance captured along their corresponding eigenvectors.

Example:
Principal Component	Eigenvalue
PC1	9
PC2	1


Here:
- PC1 captures more variance.
- PC2 captures less variance.
Therefore, PC1 is more important than PC2.

Remember:
- Eigenvector → Direction.
- Eigenvalue → Amount of variance in that direction.

# 7. PCA – Step-by-Step Solution
PCA generally follows these steps:

Step 1: Collect the Dataset

Take numerical features from the dataset.

Step 2: Standardization

Bring features onto a comparable scale.

Z = \frac{X-\mu}{\sigma}

Step 3: Calculate Covariance Matrix

Understand how features vary together.

Step 4: Calculate Eigenvalues and Eigenvectors
- Eigenvalues tell us how much variance is captured.
- Eigenvectors tell us the directions.
Step 5: Sort Eigenvalues

Arrange eigenvalues in descending order.
PC1 → Highest Eigenvalue
PC2 → Second Highest
PC3 → Third Highest

Step 6: Select Principal Components

Choose the components that preserve the required amount of variance.

Step 7: Project Data

Transform original data into the selected principal components.
Original Dataset (100 Features)
             |
             v
            PCA
             |
             v
  Reduced Dataset (10 Features)

# 11. Explained Variance
Definition: Explained Variance tells us how much of the total variance in the original dataset is captured by each principal component.

Example:
Component	Explained Variance
PC1	60%
PC2	30%
PC3	10%


Meaning:
- PC1 captures 60% information (variance).
- PC2 captures 30%.
- PC3 captures 10%.
If we select PC1 and PC2:

60\%+30\%=90\%

So, we retain 90% of the total variance.
Python Code
pca = PCA()pca.fit(X_train_scaled)print(pca.explained_variance_ratio_)


Important:

explained_variance_ratio_ tells us the proportion of total variance explained by each principal component.

# 12. Finding the Optimum Number of Principal Components
How do we decide how many components to keep?

We use Cumulative Explained Variance.

It calculates the total variance captured by multiple principal components.

Example:
Components	Cumulative Variance
PC1	60%
PC1 + PC2	90%
PC1 + PC2 + PC3	95%
PC1 + PC2 + PC3 + PC4	98%


If our target is 95%, we select 3 components.
Python Code
pca = PCA()pca.fit(X_train_scaled)import numpy as npcumulative_variance = np.cumsum(    pca.explained_variance_ratio_)print(cumulative_variance)


Automatically finding the number of components:
pca = PCA(n_components=0.95)X_train_pca = pca.fit_transform(X_train_scaled)X_test_pca = pca.transform(X_test_scaled)print(pca.n_components_)


Here, 0.95 means PCA will select enough components to preserve at least 95% of the variance.

# 13. When Does PCA Not Work Well?
PCA is useful, but it does not work perfectly in every situation.

1. Non-linear Data
- PCA captures linear directions.
- It may fail to represent complex nonlinear relationships.
2. Interpretability
- Principal components are combinations of original features.
- Understanding their real-world meaning can be difficult.
3. Important Low-Variance Features
- PCA prioritizes high variance.
- A low-variance feature may still contain important predictive information.
4. Different Feature Scales
- Features with large numerical scales may dominate PCA.
- Standardization can help.
5. Too Much Information Loss
- Selecting very few components may remove useful information.
6. Outliers
- PCA is sensitive to extreme values because covariance and variance can be affected by outliers.

# 14. Advantages of PCA
1. Reduces the number of dimensions.
2. Reduces computational complexity.
3. Helps visualize high-dimensional data.
4. Can reduce noise.
5. Can help reduce overfitting.
6. Removes linear correlation between principal components.

# 15. Disadvantages of PCA
1. Some information may be lost.
2. Principal components are difficult to interpret.
3. Sensitive to outliers.
4. Captures linear relationships.
5. Requires appropriate feature scaling.

# Important Interview Questions
Q1. What is PCA?

PCA is an unsupervised dimensionality reduction technique that transforms original features into new uncorrelated features called principal components while preserving maximum variance.

Q2. Why is variance important in PCA?

PCA uses variance to identify directions that capture the maximum spread of data.

Q3. What is the difference between Eigenvalue and Eigenvector?

Eigenvector represents the direction, while eigenvalue represents the variance captured along that direction in PCA.

Q4. What is PC1?

PC1 is the first principal component that captures the maximum possible variance.

Q5. Why do we standardize data before PCA?

To prevent features with larger numerical scales from dominating the principal components.

Q6. How do we find the optimum number of components?

Using cumulative explained variance. For example, we can select enough components to retain 95% variance.
