# Linear Regression – 5 Assumptions (Interview Cheat Sheet)
# 1. Linear Relationship
Meaning: Input features aur target variable ke beech linear relationship hona chahiye.
- X badhega toh Y ideally linear pattern follow karega.
- Relationship straight-line pattern mein honi chahiye.
- Check: Scatter Plot, Residual Plot.
Interview Answer: The relationship between independent and dependent variables should be linear.

# 2. No Multicollinearity
Meaning: Independent variables ke beech high correlation nahi honi chahiye.

Example:
- House Size
- Number of Rooms
- Number of Bedrooms
Ye features highly correlated ho sakte hain.
- Problem: Coefficients unstable ho jaate hain.
- Check: Correlation Matrix, VIF.
- Solution: Feature Removal, Regularization (Ridge/Lasso).
Interview Answer: Independent variables should not be highly correlated with each other.

# 3. Normality of Residuals
Meaning: Prediction errors (Residuals) approximately normally distributed hone chahiye.

Formula:
\[
Residual = Actual - Predicted
\]
- Check: Histogram, Q-Q Plot.
- Important: Normality residuals ki check hoti hai, input features ki nahi.
Interview Answer: Residuals should be approximately normally distributed, especially for reliable statistical inference.

# 4. Homoscedasticity
Meaning: Residuals ka variance har prediction level par approximately constant hona chahiye.

Example:
- Low predicted values → Error spread small.
- High predicted values → Error spread bhi similar hona chahiye.
- Check: Residual Plot.
- Problem: Standard errors aur statistical tests unreliable ho sakte hain.
- Solution: Log Transformation, Weighted Least Squares.
Interview Answer: The variance of residuals should remain constant across all levels of predicted values.

# 5. No Autocorrelation of Errors
Meaning: Ek observation ka error doosre observation ke error se correlated nahi hona chahiye.

Example:
- Today's stock prediction error.
- Tomorrow's stock prediction error.
Agar errors connected hain toh autocorrelation ho sakta hai.
- Check: Durbin-Watson Test.
- DW ≈ 2 → No autocorrelation.
- DW < 2 → Positive autocorrelation.
- DW > 2 → Negative autocorrelation.

### Interview Answer: Residuals should be independent of each other.
Quick Revision Table
Assumption	Simple Meaning	Check
Linear Relationship	X-Y linear relation	Scatter Plot
No Multicollinearity	Features highly correlated na hon	VIF
Normality	Residuals normally distributed	Q-Q Plot
Homoscedasticity	Constant error variance	Residual Plot
No Autocorrelation	Errors independent hon	Durbin-Watson


# Interview: 5 Important Questions
Q1. What happens if multicollinearity exists?

Ans: Coefficients become unstable and difficult to interpret.

Q2. What is a residual?

Ans: Difference between actual and predicted values.

Q3. Does Linear Regression require normally distributed input features?

Ans: No, normality assumption applies to residuals, not input features.

Q4. What is heteroscedasticity?

Ans: When residual variance is not constant.

Q5. Which test checks autocorrelation?

Ans: Durbin-Watson test.

# Memory Trick 🧠
L – M – N – H – A
1. L – Linear Relationship
2. M – Multicollinearity (No)
3. N – Normality
4. H – Homoscedasticity
5. A – Autocorrelation (No)
Interview Golden Point: Linear Regression ke liye normality of residuals aur homoscedasticity mainly statistical inference (p-values, confidence intervals) ke liye important hain. Accurate prediction ke liye har assumption equally mandatory nahi hoti.
