# Batch (Offline) vs Online Machine Learning


1. Offline Learning (Batch ML)
Model is trained on the available dataset at once and updated through periodic retraining.

2. Online Learning
Model learns continuously or incrementally from incoming data.

### Difference Between Offline & Online Learning
Features	        Offline Learning	                                    Online Learning
1. Complexity	        Less complex; model remains constant after training.More complex; model keeps evolving over time.
2. Computational Power	Fewer computations; one-time batch training.	    Continuous computations for model updates.
3. Use in Production	Easier to implement and manage.	                    Difficult to implement and manage.
4. Applications	Image   classification and stable data patterns.	        Finance, economics and healthcare where new patterns emerge.
5. Tools	           Scikit-learn, TensorFlow.	                        MOA (Massive Online Analysis) and other streaming ML tools.

I# mportant Concepts
## 1. When to Use Online Learning?

When data arrives continuously and patterns change frequently.

Example: Stock market prediction, fraud detection.

## 2. How to Implement Online Learning?
Collect incoming data.
Train/update the model incrementally.
Monitor model performance.
Repeat the process.

## 3. Learning Rate

Learning rate controls how much the model updates its parameters when learning new data.

High learning rate = Learns quickly but may forget old patterns.
Low learning rate = Learns slowly but updates gradually.

## 4. Out-of-Core Learning

A technique used to train ML models when the dataset is too large to fit into RAM.

Example: Training a model on a 100 GB dataset using a computer with 8 GB RAM by processing data in small batches.

## 5. Disadvantages of Online Learning
Requires continuous monitoring.
Sensitive to bad or noisy data.
May forget previously learned patterns (Catastrophic Forgetting).
More difficult to implement and manage.
