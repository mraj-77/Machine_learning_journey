# Business Problem → ML Problem (Netflix): Medium Summary
Core Idea

Business goal: Revenue badhana
Business metric: Churn rate kam karna (4% → 3.75% ya 3.5%)
ML problem: Har user ke liye predict karo ki wo churn karega ya nahi (Binary Classification)

Churn ka matlab: user subscription cancel karke chala jaata hai. Chhota sa farak bhi crores users par bahut bada revenue impact deta hai.

7 Steps

### 1. Business Problem to ML Problem
Revenue badhao → Churn rate kam karo → Churn predict karo → At-risk users ko retention offer do.

### 2. Type of Problem
Classification (Churn = 1, No churn = 0). Output probability bhi ho sakta hai, jaise 0.8 matlab 80% chance ki user jaayega.

### 3. Current Solution
Abhi company is problem ko kaise handle karti hai? Ye benchmark banta hai. ML model tabhi useful hai jab wo current solution se behtar ho.

### 4. Getting Data

Watch time: Kam hota watch time = disengagement
Search but did not find: Dhundha par nahi mila = frustration
Content left in the middle: Beech mein chhoda = galat recommendation ya bura experience
Clicked on recommendations (order ke saath): Top par click hua ya neeche, isse recommendation quality pata chalti hai

Ye sab implicit feedback hai: user ne bola nahi, uske behavior se andaza lagta hai.

### 5. Metrics to Measure
Precision, Recall, F1, AUC. Churn mein Recall important hai, kyunki jaane wale user ko miss karna costly hai.

### 6. Online vs Batch
Prediction real-time chahiye (user ke login par) ya roz raat ko batch mein? Churn ke liye usually batch kaafi hota hai.

### 7. Check Assumptions
Kya hamari assumptions sahi hain? Jaise: “kam watch time = churn” sach hai ya nahi, aur data har type ke user ko represent karta hai ya nahi.

Final Flow

Revenue ↑ → Churn ↓ → User behavior data → Churn prediction model → Retention offers → Churn kam
