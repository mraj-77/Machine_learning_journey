# Scikit-Learn Pipelines
## 1. Definition

A Pipeline is a sequence of multiple steps in which the output of one step becomes the input of the next step.

## 2. Why Use Pipelines?
Combines multiple preprocessing and ML steps.
Applies the same preprocessing to training and testing data.
Reduces code repetition.
Helps prevent data leakage.
Makes code clean and organized.

## 3. Example

Suppose we have a dataset with missing values and categorical data.

Pipeline steps:

Imputation: Handle missing values.
Encoding: Convert categorical data into numerical data.
Scaling: Bring features to a similar scale.
Model Training: Train the Machine Learning model.

Flow:

Raw Data → Imputation → Encoding → Scaling → ML Model → Prediction

## 4. Simple Python Example
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

pipe = Pipeline([
    ('imputer', SimpleImputer(strategy='mean')),
    ('scaler', StandardScaler()),
    ('model', LogisticRegression())
])

pipe.fit(X_train, y_train)

y_pred = pipe.predict(X_test)

## 5. Important Functions
Function	Use
fit()	Learn from training data
transform()	Transform data
fit_transform()	Fit and transform together
predict()	Make predictions

Remember: Pipeline connects multiple ML steps together and makes preprocessing easier and consistent.

Real-life example: Just like a factory where raw materials pass through different machines one by one to produce a final product, a Pipeline processes data through multiple steps to generate predictions.

# ColumnTransformer = alag columns pe alag transformation lagata hai.
Jaise: Age pe mean imputation, Sex/Embarked pe OHE.

### Chhota example:

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.tree import DecisionTreeClassifier

ct = ColumnTransformer([
    ('age', SimpleImputer(strategy='mean'), ['Age']),
    ('emb', Pipeline([
        ('imp', SimpleImputer(strategy='most_frequent')),
        ('ohe', OneHotEncoder(handle_unknown='ignore'))
    ]), ['Embarked']),
    ('sex', OneHotEncoder(handle_unknown='ignore'), ['Sex'])
])

pipe = Pipeline([
    ('prep', ct),
    ('model', DecisionTreeClassifier())
])

pipe.fit(X_train, y_train)
pipe.predict(X_test)
