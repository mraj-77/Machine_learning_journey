# 1. Introduction to Linear Regression

**Linear Regression** is a Supervised Machine Learning algorithm used to predict a continuous numerical value based on one or more input features.

### Simple Definition
Linear Regression finds the relationship between input (X) and output (Y) and uses that relationship to make predictions.

### Example
Suppose we have student data:

| CGPA (Input) | Package (Output) |
|---|---:|
| 6.0 | 3 LPA |
| 7.0 | 4 LPA |
| 8.0 | 6 LPA |
| 9.0 | 8 LPA |

Our goal is to predict a student's package based on CGPA.

For example:
- Input: CGPA = 8.5
- Output: Predicted Package = 7 LPA (hypothetically)

### Why Linear Regression?
- Easy to understand.
- Easy to implement.
- Easy to interpret.
- Works well when the relationship between input and output is approximately linear.
- Useful for numerical predictions.

# 3. Types of Machine Learning

Machine Learning is broadly divided into two major categories:

### A. Supervised Learning

In Supervised Learning, the model learns from labelled data.

Labelled Data means we have both:
- Input (Features)
- Output (Target)

Example:

| CGPA (X) | Package (Y) |
|---|---|
| 6.0 | 3 LPA |
| 7.0 | 4 LPA |
| 8.0 | 6 LPA |

The model learns the relationship between CGPA and Package.

**Types of Supervised Learning:**

1. Regression
2. Classification

**Regression:** Predicts numerical values.

Example: Predicting house prices.

**Classification:** Predicts categories or classes.

Example:
- Email → Spam
- Email → Not Spam

### B. Unsupervised Learning

In Unsupervised Learning, the model learns from unlabelled data.

Here, we don't provide the correct output (Target) to the model.

Example:

Suppose we have customer data:

| Age | Income |
|---|---|
| 22 | 20,000 |
| 25 | 25,000 |
| 45 | 80,000 |
| 50 | 90,000 |

The algorithm can identify different customer groups automatically.

**Common Techniques:**
1. Clustering
2. Dimensionality Reduction
3. Association Rule Learning

**Remember:**

| Supervised Learning | Unsupervised Learning |
|---|---|
| Labelled Data | Unlabelled Data |
| Input + Output | Only Input |
| Regression | Clustering |
| Classification | Dimensionality Reduction |

**Important:** Linear Regression belongs to Supervised Learning → Regression.

# 4. Types of Linear Regression

Linear Regression can be divided into three commonly studied forms:

1. Simple Linear Regression
2. Multiple Linear Regression
3. Polynomial Regression

## A. Simple Linear Regression (SLR)

Simple Linear Regression uses:
- One Input Feature (X)
- One Output Feature (Y)

Example:

Predicting a student's package using only CGPA.

| CGPA (X) | Package (Y) |
|---|---|
| 6.0 | 3 LPA |
| 7.0 | 4 LPA |
| 8.0 | 6 LPA |
| 9.0 | 8 LPA |

Only one input column (CGPA) is used to predict one output column (Package).

### Formula

y = mx + c

Where:

- y = Predicted Output
- x = Input Feature
- m = Slope (Coefficient)
- c = Intercept

### Example

y = 2x + 1

If x = 5:

y = 2(5) + 1

y = 11

Therefore, predicted output = 11.

### Graph

Simple Linear Regression represents a straight line.

The model tries to find the best-fit line through the data points.

---

## B. Multiple Linear Regression (MLR)

Multiple Linear Regression uses:
- Multiple Input Features (X)
- One Output Feature (Y)

Example:

Predicting a student's package using:

1. CGPA
2. Gender
3. 12th Marks
4. State

| CGPA | Gender | 12th Marks | State | Package |
|---|---|---|---|---|
| 8.0 | Male | 85 | Punjab | 6 LPA |
| 7.0 | Female | 90 | Bihar | 5 LPA |
| 9.0 | Male | 95 | Delhi | 9 LPA |

Here:

**Input Features (X):**
- CGPA
- Gender
- 12th Marks
- State

**Output / Target (Y):**
- Package

The model uses all input features to predict one output.

### Formula

y = b0 + b1x1 + b2x2 + b3x3 + ... + bnxn

Where:

- y = Predicted Output
- b0 = Intercept
- b1, b2, b3 = Coefficients
- x1, x2, x3 = Input Features
- n = Number of Input Features

### Example

Package = b0 + b1(CGPA) + b2(Gender) + b3(12th Marks) + b4(State)

Note: Categorical columns such as Gender and State must be encoded into numerical values before using Linear Regression.

---

## C. Polynomial Regression

Polynomial Regression is used when the relationship between input and output is curved rather than approximately straight.

Example:

Suppose we want to predict salary based on experience.

The relationship might not be a straight line.

### Formula

y = b0 + b1x + b2x² + b3x³ + ... + bnxⁿ

Where:

- x = Input Feature
- x² = Square of Input
- x³ = Cube of Input
- b0, b1, b2 = Model Coefficients

### Example

y = 2 + 3x + 4x²

If x = 2:

y = 2 + 3(2) + 4(2²)

y = 2 + 6 + 16

y = 24

**Important:** Polynomial Regression is still a linear regression model in terms of its coefficients, even though it can represent a curved relationship.

# 5. Difference Between SLR, MLR and Polynomial Regression

| Feature | Simple LR | Multiple LR | Polynomial LR |
|---|---|---|---|
| Full Form | Simple Linear Regression | Multiple Linear Regression | Polynomial Regression |
| Input Features | One | Multiple | One or Multiple |
| Output | One | One | One |
| Relationship | Straight line | Linear combination of features | Can represent curves |
| Formula | y = mx + c | y = b0 + b1x1 + b2x2 | y = b0 + b1x + b2x² |

# 7. Python Implementation of Linear Regression

### Step 1: Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
```

### Step 2: Create Dataset

```python
data = {
    'CGPA': [6, 7, 8, 9, 10],
    'Package': [3, 4, 6, 8, 10]
}

df = pd.DataFrame(data)
```

### Step 3: Separate Input and Output

```python
X = df[['CGPA']]

y = df['Package']
```

### Step 4: Create Model

```python
model = LinearRegression()
```

### Step 5: Train Model

```python
model.fit(X, y)
```

### Step 6: Make Prediction

```python
model.predict([[8.5]])
```

The model predicts the package for a student with CGPA 8.5.

### Step 7: Check Coefficient and Intercept

```python
print(model.coef_)

print(model.intercept_)
```

- `coef_` → Slope
- `intercept_` → Intercept

Suppose hum experience ke basis par employee ki salary predict kar rahe hain.

Formula:

y = mx+b

- y = Salary

- x = Experience
- b = Base Salary (Intercept)
Case 1: Agar b = 0 ho

Maan lo:
- m = 5,000
- Experience = 0 years
- b = 0
Salary = 5,000 × 0 + 0 = ₹0

Lekin real life mein fresher ko bhi salary mil sakti hai, jaise ₹15,000.

Case 2: Agar b = 15,000 ho

Salary = 5,000 × 0 + 15,000 = ₹15,000

Ab prediction zyada realistic hai.

Conclusion: Intercept (b) model ko baseline value provide karta hai. Lekin mathematically b ka zero hona impossible nahi hai. Model training ke dauran data ke according b zero bhi ho sakta hai aur non-zero bhi.

Yaad rakho: Intercept = Jab X = 0 ho, tab model ki predicted Y value.
