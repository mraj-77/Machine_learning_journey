# Ridge Regression – 5 Key Understandings

### 1. How Coefficients Get Affected
- Ridge Regression shrinks model coefficients towards **zero**.
- Coefficients become smaller but generally not exactly zero.
- Higher λ → More coefficient shrinkage.

### 2. Higher Values Are Impacted More
- Ridge uses the square of coefficients (L2).
- Large coefficients receive a larger penalty.
- Example:
  - 2² = 4
  - 5² = 25
  - 10² = 100

**Remember:** Larger coefficients receive a stronger penalty.

### 3. Bias-Variance Tradeoff
- Regularization increases Bias.
- Regularization decreases Variance.
- Helps prevent Overfitting.

| Lambda | Bias | Variance |
|---|---|---|
| Low λ | Low | High |
| High λ | High | Low |

### 4. Contour Plots
- Used to visualize Ridge Regression optimization.
- Circular contours represent L2 penalty.
- Best solution is found where the loss contour touches the constraint boundary.
- Ridge pulls coefficients towards zero.

### 5. Why Called Ridge?
- Ridge Regression was named after the ridge-like solutions produced in regression with multicollinearity.
- It helps handle highly correlated features.
- Makes coefficient estimates more stable.

### Practical Tips
- Always apply Feature Scaling before Ridge.
- Use `StandardScaler` for standardization.
- Tune `alpha` using Cross-Validation.
- Use Ridge when features are highly correlated.
- Higher alpha means stronger regularization.
- It is apply when more than 2 input feature
