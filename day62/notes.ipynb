# Bias-Variance Trade-off
1. Bias kya hai?
Model ki prediction aur actual value ke beech systematic error.

2. Variance kya hai?
Training data change hone par model ki prediction kitni change hoti hai.

3. Three Models (Most Important)
Model	Bias	Variance	Problem
M1 – Simple	High	Low	Underfitting
M2 – Balanced	Low	Low	Good Model ✅
M3 – Complex	Low	High	Overfitting


M1 – Underfitting
- Model bahut simple hai.
- Data ke patterns learn nahi kar pata.
- High Bias + Low Variance.
- Training aur testing dono par poor performance.

M2 – Perfect Fit
- Model patterns ko properly learn karta hai.
- Low Bias + Low Variance.
- Training aur testing dono par good performance.

M3 – Overfitting
- Model bahut complex hai.
- Training data ko memorize kar leta hai.
- Low Bias + High Variance.
- Training accuracy high, testing accuracy low.

4. Overfitting ko kaise control karein?
Image mein diye 3 techniques:
1. Regularization – Model ki complexity control karta hai.
2. Bagging – Multiple models combine karke variance reduce karne mein help karta hai.
3. Boosting – Models sequentially train karke errors improve karta hai; often bias reduce karne mein help karta hai.
5. Interview Cheat Sheet ⭐
Question	Answer
High Bias?	Underfitting
High Variance?	Overfitting
Low Bias + Low Variance?	Ideal Model
High Bias + Low Variance?	Model too simple
Low Bias + High Variance?	Model too complex
Overfitting solution?	Regularization, Bagging
Underfitting solution?	Increase model complexity


🧠 Yaad rakhne ki Trick
- High Bias = Model simple = Underfitting
- High Variance = Model complex = Overfitting
- Low Bias + Low Variance = Best Model
One-line definition (Interview):

Bias-Variance Trade-off is the balance between underfitting and overfitting to achieve good performance on unseen data.
