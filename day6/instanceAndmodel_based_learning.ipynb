# 1. Instance Vs Model Based Learning (Introduction)

Ye pehla section hai, jo aapki image mein hai. Basic idea: learning do tarah ki hoti hai.

Memorizing → examples yaad karna
Generalizing → rule/pattern seekhna

# 2. Instance Based Learning

System training data ko yaad rakhta hai. Naya data aane par wo purane examples se similarity (kitna milta-julta hai) check karta hai aur uske basis par answer deta hai.

Example: Aap ek naya movie dekhne wale ho. System dekhta hai ki aapke jaise taste wale logon ne kya pasand kiya, aur wahi suggest kar deta hai.

Algorithm: KNN
Fayda: Simple hai, training time kam
Nuksan: Saara data store karna padta hai, prediction slow ho sakta hai

# 3. Model Based Learning

System data se pattern seekhkar ek model (formula) banata hai. Phir naya data aane par sirf model use hota hai.

Example: Ghar ki size se price predict karna. System seekh leta hai ki “size badhne par price badhta hai” aur ek formula bana leta hai, jaise:

Price = 5000 × Size + 20000

Ab koi bhi naye ghar ki size daalo, price mil jayega.

Algorithm: Linear Regression, Logistic Regression
Fayda: Prediction fast, data store karne ki zaroorat nahi
Nuksan: Agar model galat bana, to predictions bhi galat honge

# 4. Differences

Point	            Instance Based	                Model Based
Seekhne ka tarika	Yaad karta hai (memorize)	Pattern seekhta hai (generalize)
Training data	    Hamesha chahiye	            Training ke baad nahi chahiye
Prediction speed	Slow	                    Fast
Training time	    Kam                     	Zyada
Memory use	        Zyada	                    Kam
Example	            KNN	                        Linear 

Ek Line Mein Yaad Rakho
Instance Based = “Mujhe yaad hai, maine pehle aisa kuch dekha tha” 🧠
Model Based = “Mujhe rule pata hai, isliye main bata sakta hoon” 📏
