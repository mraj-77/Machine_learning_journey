# 1. What are Tensors?

Tensor ek multi-dimensional array hai jisme numbers store hote hain. Computer text, image, sound sab kuch numbers mein badalkar tensor mein rakhta hai. TensorFlow aur PyTorch jaisi libraries isi par kaam karti hain.

# 2. 0D Tensor / Scalar

Sirf ek single number. Koi list nahi, koi direction nahi.
Example: 5, 3.14, ya kisi ki age = 25

# 3. 1D Tensor / Vector

Numbers ki ek line (list).

Example: [25, 60, 170] (age, weight, height)

Dimension (Rank) = 1, kyunki sirf ek axis hai (ek hi direction mein numbers hain).
Vector ki size = 3, kyunki isme 3 numbers hain.
Shape = (3,)

# 4. 2D Tensor / Matrix

Numbers ki table (rows aur columns).
Example:

[[25, 60, 170],
 [30, 75, 180]]

Excel sheet ya CSV file isi tarah ki hoti hai.

# 5. ND Tensors

Jab dimensions 3, 4, 5 ya usse zyada ho jayein, to use N-Dimensional tensor kehte hain.

# 6. Rank, Axes and Shape

Ye 3 terms confuse karte hain, to simple rakho:

Term	Matlab	Example [[1,2,3],[4,5,6]]
Rank	Kitne dimensions hain	2
Axes	Dimensions ki directions (axis 0, axis 1...)	2 axes
Shape	Har axis mein kitne elements hain	(2, 3)

Yahan shape (2, 3) ka matlab: 2 rows, 3 columns.

no. pf axis = no. of ndim = rank = no. of row

# 7. Example of 1D Tensors

Ek student ke marks: [80, 90, 75]
Ya ek ghar ki details: [size, rooms, price]. Shape: (3,)

# 8. Example of 2D Tensors

Poori class ka data (table): har row = ek student, har column = ek feature.
Ye ML ka sabse common format hai. Shape: (students, features), jaise (100, 5).

# 9. Example of 3D Tensors

Time series ya text data. Jaise 100 logon ka 7 din ka, har din 3 readings ka data.
Shape: (100, 7, 3)
Ya ek colour image: (height, width, 3) jahan 3 = Red, Green, Blue channels.

# 10. Example of 4D Tensors

Images ka batch. Jaise 32 colour images ek saath.
Shape: (32, 224, 224, 3) = (kitni images, height, width, channels)

# 11. Example of 5D Tensors

Video data. Video = images ki sequence, to ek extra dimension aata hai.
Shape: (batch, frames, height, width, channels), jaise (4, 30, 224, 224, 3) = 4 videos, har ek mein 30 frames.

Quick Summary Table
Dimension	Naam	Example
0D	Scalar	Ek number (5)
1D	Vector	List ([1,2,3])
2D	Matrix	Table / Excel sheet
3D	3D Tensor	Colour image, time series
4D	4D Tensor	Images ka batch
5D	5D Tensor	Videos ka batch
