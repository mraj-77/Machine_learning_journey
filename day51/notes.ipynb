# how to find m and b

# 1. Closed Form Solution (OLS)
OLS = Ordinary Least Squares.

Is method mein mathematical formula use karke directly best-fit line ke m aur b find karte hain.

Example:
CGPA (X)	Package (Y)
1	3
2	5
3	7


Humein aisi line find karni hai jo data ko best fit kare.

OLS se:
- m = 2
- b = 1
Ab equation:
\[
y = 2x+1
\]
Agar X = 4:

Y = 2(4) + 1 = 9

Predicted output = 9.

# 2. Non-Closed Form Solution (Gradient Descent)
Is method mein m aur b directly calculate nahi hote.

Algorithm:
1. Initially m aur b ki values leta hai.
2. Prediction karta hai.
3. Error calculate karta hai.
4. m aur b ko update karta hai.
5. Ye process repeat karta hai jab tak error minimum na ho jaye.
Difference
OLS	Gradient Descent
Direct mathematical solution	Iterative solution
Closed-form method	Optimization method
Matrix calculations	Repeated parameter updates
Small/medium datasets mein useful	Large datasets mein useful


Yaad rakhna:
- OLS → Directly m and b find karta hai.
- Gradient Descent → Step-by-step m and b optimize karta hai.
Dono ka objective same hai: Best Fit Line find karna by minimizing prediction error.
