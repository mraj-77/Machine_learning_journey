# 1. Handling Missing Numerical Data
The process of identifying and filling missing numerical values (NaN) in a dataset.

It has two main approaches:
- Univariate Imputation: Uses a single column to fill missing values.
- Multivariate Imputation: Uses multiple columns to predict missing values.

# 2. Mean/Median Imputation
Replaces missing numerical values with the mean or median of the column.
- Mean: Average of all values.
- Median: Middle value of sorted data.
Python:
df['Age'].fillna(df['Age'].mean())df['Age'].fillna(df['Age'].median())

1. Advantages
1. Easy to Implement: Simple to understand and apply.
2. Fast: Requires less computational time.
3. Preserves Dataset Size: Does not remove rows.
4. Works with Small Datasets: Useful when only a few values are missing.
5. Median Handles Outliers: Median is less affected by extreme values.

2. Disadvantages
1. Distortion of Distribution: Changes the original data distribution.
2. Loss of Variance: Reduces variation in the dataset.
3. Creates Bias: Can introduce bias in data.
4. Ignores Relationships: Does not consider relationships between different features.
5. Not Suitable for High Missing Data: May produce unreliable results when many values are missing.

less than 5% missing data.


# 3. Arbitrary Value Imputation
Replaces missing values with a manually selected constant value, such as 0, -1 or 999.

Python:
df['Age'].fillna(0)

# 4. End of Distribution Imputation
Replaces missing values with an extreme value at the end of the distribution.

Generally uses values far below or above the normal data range.

Python:
df['Age'].fillna(df['Age'].mean() + 3 * df['Age'].std())


# 5. Random Sample Imputation
Replaces missing values with randomly selected existing values from the same column.

Python:
df['Age'].fillna(    df['Age'].dropna().sample(df['Age'].isnull().sum()).values)


# 6. Automatically Select Best Imputation
Uses statistical methods or machine learning techniques to select a suitable imputation strategy.

Examples:
- Mean/Median Imputation
- KNN Imputation
- Iterative Imputation (MICE)
Python:
from sklearn.impute import SimpleImputerimputer = SimpleImputer(strategy='median')df['Age'] = imputer.fit_transform(df[['Age']]).ravel()
