# Regression Metrics
# 1. MAE (Mean Absolute Error)
Meaning: Actual value aur Predicted value ke beech average absolute difference.

Formula:
\[
MAE = \frac{1}{n}\sum |y_i-\hat y_i|
\]
Example:
Actual	Predicted	Error
10	12	2
20	18	2
30	33	3


MAE = (2 + 2 + 3) / 3

MAE = 2.33

Intuition: Model average 2.33 units ki galti kar raha hai.
- Lower MAE = Better Model
- MAE = 0 means perfect prediction.

# 2. MSE (Mean Squared Error)
Meaning: Errors ka square karke unka average calculate karna.

Formula:
\[
MSE = \frac{1}{n}\sum(y_i-\hat y_i)^2
\]
Example:
Actual	Predicted	Error	Error²
10	12	2	4
20	18	2	4
30	33	3	9


MSE = (4 + 4 + 9) / 3

MSE = 5.67

Intuition: Error ko square karne se large errors ko zyada punishment milti hai.
- Lower MSE = Better Model.
- MSE = 0 means perfect prediction.

# 3. RMSE (Root Mean Squared Error)
Meaning: MSE ka square root lena.

Formula:
\[
RMSE = \sqrt{MSE}
\]
Example:

MSE = 5.67

RMSE = √5.67

RMSE = 2.38

Intuition: RMSE bhi prediction error batata hai, lekin ye original data ki same unit mein hota hai.
- Lower RMSE = Better Model.
- Large errors ko zyada importance deta hai.

# 4. R² Score (R-Squared)
Meaning: R² Score batata hai ki model target variable ke variation ko kitna explain kar pa raha hai.

Formula:
\[
R^2 = 1-\frac{SS_{res}}{SS_{tot}}
\]
Example:

Suppose hum house prices predict kar rahe hain.
- R² = 1 → Perfect prediction.
- R² = 0.80 → Model 80% variation explain karta hai.
- R² = 0 → Mean prediction jitna performance.
- R² < 0 → Mean prediction se bhi worse performance.
Intuition: Higher R² generally means better model.

# 5. Adjusted R² Score
Meaning: Adjusted R², R² Score ka improved version hai jo unnecessary features add karne par penalty deta hai.

Example:

Suppose tum house price predict kar rahe ho.

Initially features:
- Area
- Location
- Bedrooms
Ab tum unnecessary feature add kar dete ho:
- House ka colour
Normal R² kabhi-kabhi unnecessary feature add karne par bhi increase ho sakta hai.

Lekin Adjusted R² unnecessary features ko penalize karta hai.

Formula:
\[
Adjusted\ R^2 =
1-\frac{(1-R^2)(n-1)}{n-k-1}
\]
Where:
- n = Number of observations.
- k = Number of input features.
Intuition: Ye check karta hai ki naye features model ko actually improve kar rahe hain ya nahi.

# 6. All Metrics – Quick Comparison
Metrics	Full Form	Better Value
MAE	Mean Absolute Error	Lower
MSE	Mean Squared Error	Lower
RMSE	Root Mean Squared Error	Lower
R²	R-Squared	Higher
Adjusted R²	Adjusted R-Squared	Higher

# 8. One-Minute Revision
Metric	Yaad Rakhne Ka Tarika
MAE	Average absolute error
MSE	Squared error ka average
RMSE	MSE ka square root
R²	Model kitna variation explain karta hai
Adjusted R²	Unnecessary features par penalty


Interview Trick
- MAE vs MSE: MSE large errors ko zyada punish karta hai.
- MSE vs RMSE: RMSE original unit mein hota hai.
- R² vs Adjusted R²: Adjusted R² features ki unnecessary addition ko penalize karta hai.
Golden Rule:

Error Metrics (MAE, MSE, RMSE) → Lower is Better.

Goodness-of-fit Metrics (R², Adjusted R²) → Higher is Better.
